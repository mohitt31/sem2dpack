# SEM2DPACK AVX2 Microbenchmark (Colab)
Run this notebook on a Google Colab instance to get genuine x86-64 AVX2 timings.

```
# 1. Confirm real x86 + AVX2
!uname -m
!grep -m1 -o avx2 /proc/cpuinfo || echo "NO AVX2 - runtime change karo"
```

```
# 2. Pull the branch
!rm -rf sem2dpack && git clone -q --branch bench/kernel-microbench https://github.com/mohitt31/sem2dpack.git
%cd sem2dpack
```

```
# 3. Find the microbench + H file (adjust path if needed)
!find . -name "microbench.cpp" -o -name "H_ngll5.txt" | head
```

```
# 4. Build with native AVX2 (NO -ffast-math) and confirm AVX2 emitted
!g++ -O3 -march=native -std=c++17 $(find . -name microbench.cpp | head -1) -o microbench 2>&1 | head
!objdump -d microbench | grep -m3 -oE "vfmadd[0-9a-z]*|ymm[0-9]+"
```

```
# 5. Run pinned, median-of-3 (H_ngll5.txt must be in the run dir)
!cp $(find . -name H_ngll5.txt | head -1) . 2>/dev/null; taskset -c 0 ./microbench 2000
```
